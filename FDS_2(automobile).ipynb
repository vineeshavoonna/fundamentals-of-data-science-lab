```python
#euclidian distance
temp = df[['city-mpg','engine-size']].dropna()
ez= temp.iloc[1]['engine-size']
cm= temp.iloc[1]['city-mpg']
ez2= temp.iloc[2]['engine-size']
cm2= temp.iloc[2]['city-mpg']
distance = np.sqrt((ez-ez2)**2 + (cm-cm2)** 2)
print("\nEuclidean Distance")
print(distance)
```

```python
#manhattan distance
manhattan = abs(ez-ez2)+abs(cm-cm2)
print("\nManhattan Distance")
print(manhattan)
```

```python
#minkowski distance
minkowski=((abs(ez-ez2)**3)+(abs(cm-cm2)**3))**(1/3)
print("\nMinkowski Distance")
print(minkowski)
```

```python
#cosine dissimilarity
s=np.array([ez,cm])
v=np.array([ez2,cm2])
cosine_similarity=np.dot(s,v)/(np.linalg.norm(s)*np.linalg.norm(v))
cosine_dissimilarity=1-cosine_similarity
print("\nCosine Dissimilarity")
print(cosine_dissimilarity)
```
