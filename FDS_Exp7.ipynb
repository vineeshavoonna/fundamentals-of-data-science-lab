```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    'Student_ID': [1, 2, 3, 4],
    'Name': ['Sai', 'Ravi', 'Priya', 'Kiran'],
    'Department': ['CSE', 'ECE', 'CSE', 'IT'],
    'Marks': [85, 78, 92, 74]
})

df2 = pd.DataFrame({
    'Student_ID': [1, 2, 3, 5],
    'Attendance': [90, 85, 95, 80]
})

print('Student Details:')
display(df1)

print('Attendance Details:')
display(df2)
```

```python
multi_df = df1.set_index(['Department', 'Student_ID'])
display(multi_df)
```

```python
multi_df.loc['CSE']
```

```python
df3 = pd.DataFrame({
    'Student_ID': [6, 7],
    'Name': ['Anu', 'Manoj'],
    'Department': ['ECE', 'IT'],
    'Marks': [88, 81]
})

combined_df = pd.concat([df1, df3], ignore_index=True)
display(combined_df)
```

```python
merged_df = pd.merge(df1, df2, on='Student_ID', how='inner')
display(merged_df)
```

```python
left_merged_df = pd.merge(df1, df2, on='Student_ID', how='left')
display(left_merged_df)
```

```python
long_df = merged_df.set_index(
    ['Student_ID', 'Name', 'Department'])[['Marks', 'Attendance']].stack().reset_index()

long_df.columns = ['Student_ID', 'Name', 'Department', 'Type', 'Value']
display(long_df)
```

```python
pivot_df = long_df.pivot(
    index='Name',
    columns='Type',
    values='Value'
)
display(pivot_df)
```

```python
pivot_table_df = pd.pivot_table(
    df1,
    values='Marks',
    index='Department',
    aggfunc='mean'
)
display(pivot_table_df)
```

```python
stacked_df = pivot_df.stack()
print('After stack():')
display(stacked_df)

unstacked_df = stacked_df.unstack()
print('After unstack():')
display(unstacked_df)
```
