```python
import pandas as pd
df = pd.read_csv("/content/Iris-Dataset.csv")
print("Initial first 5 records : \n")
print(df.head())

df = df.drop(columns=["Id"])
print("\nAfter dropping column : \n")
print(df.head())

numeric_df = df.select_dtypes(include=['number'])

mean = numeric_df.mean()
median = numeric_df.median()
mode = numeric_df.mode().iloc[0]
range_val = numeric_df.max() - numeric_df.min()
variance = numeric_df.var()
std_dev = numeric_df.std()

result = pd.DataFrame({
    "Mean": mean,
    "Median": median,
    "Mode": mode,
    "Range": range_val,
    "Variance": variance,
    "Standard Deviation": std_dev
})

print("\nDescriptive Statistics of Iris Dataset : \n")
print(result.round(3))
```
